¿Qué es el método de Newton?

Es un algoritmo para encontrar raíces de funciones (valores que hacen que la función sea igual a cero), usando una fórmula iterativa:

$\text{next\_guess} = \text{guess} - \frac{f(\text{guess})}{f’(\text{guess})}$

Objetivo:

Encontrar un número g tal que:

$g^3 = x \quad \Rightarrow \quad f(g) = g^3 - x = 0$

Esto significa: hallar la raíz cúbica de x, usando el método de Newton.

¿Cómo se aplica Newton en este caso?

- La función es:
$f(g) = g^3 - x$
- Su derivada es:
$f’(g) = 3g^2$
- Entonces la fórmula de Newton se convierte en:
$\text{next\_g} = g - \frac{g^3 - x}{3g^2}$


```python
x = int(input('What x to find the cube root of? '))
g = int(input('What guess to start with? '))

print('Current estimate cubed = ', g**3)

next_g = g - ((g**3 - x) / (3*g**2))

print('Next guess to try = ', next_g)
```
